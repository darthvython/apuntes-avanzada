# Semana 3: Estructuras de datos

- **Estructura de Datos**: Construcción que permite agrupar datos relacionados junto con operaciones para accederlos y modificarlos de manera eficiente.
- **Iterar / Iterable**: Iterar significa *recorrer* una estructura. Un objeto es iterable si *puede ser recorrido*.

### Clasificación de estructuras
1. **Estructuras Secuenciales**:
   - Basadas en un ordenamiento secuencial (depende de cómo se ingresan los elementos).
   - Se pueden recorrer manteniendo el orden que induce la estructura y se puede acceder al $n$-ésimo elemento.
   - *Ejemplos*: `tuple`, `list`, `str`, stacks (pilas) y colas.
2. **Estructuras No Secuenciales**:
   - Almacenan datos sin establecer un orden fijo de acceso.
   - Proveen métodos muy eficientes para la búsqueda de datos.
   - *Ejemplos*: `dict`, `set`, árboles, grafos, heaps.

## 1. Tuplas (`tuple`)

- Estructura de datos **ordenada** e **inmutable** (no se pueden modificar, agregar ni eliminar sus elementos una vez creada).
- Son **heterogéneas**: pueden contener distintas clases o tipos de datos (incluso otras estructuras).
- **Inmutabilidad y Hasheabilidad**:
  - Los objetos inmutables son `hasheables` (permiten calcular un valor `hash` único a partir de su contenido).
  - Si una tupla contiene un elemento mutable (como una lista), deja de ser `hasheable`.

```python
tupla = (1,2,3)
# Creamos una tupla con los elementos 1,2,3
print(f"{tupla}\n")

tupla = 1, True, ("Es 15 de Agosto", "Ayuda", "Retiro ramo?")
# Creamos una tupla sin utilizar paréntesis (son opcionales)
print(f"{tupla}\n")

tupla = ("Grabándote:)",)
# Para crear una tupla con un solo elemento, se debe agregar
# una "," al final, para que no haya error de interpretación.
print(f"{tupla}\n")

tupla = ("Grabándote:)")
print(f"{tupla}\n")
# Sin la coma final

tupla = ()
print(f"{tupla}")
print(type(tupla))
# Vacía
```

#### Verifiquemos la obtención del `hash` e intentemos modificar un elemento de la tupla.

```python
tupla_1 = (1, 2, 3, 4, 5, 6, 7)
tupla_2 = (1, 2, 3, 4, 5, 6, 7)

print(tupla_1)
print(tupla_2)
print()
print(tupla_1.__hash__())
print(tupla_2.__hash__())
# Imprimimos los hash asociados a cada tupla, y como contienen a los mismos elementos,
# tienen el mismo hash.

tupla = (2.4, 1.3, False, "xd")

tupla[1] = True

# Intentamos modificar un elemento de la tupla
```

#### Desempaquetado de Elementos y Operador Asterisco (`*`)
- Las funciones pueden retornar múltiples valores empaquetados en una tupla.
- El operador `*` permite desempaquetar secuencias en variables individuales o al construir nuevas estructuras.

```python
from typing import Tuple
# 
def partes(num : float) -> Tuple[int, float]:
    parte_entera = int(num // 1)
    parte_decimal = num - parte_entera
    return parte_entera, parte_decimal

print(partes(2.45))
print()
# La función partes nos devuelve una tupla con la parte entera y decimal del float ingresado.

datos = ('Inicio', 10, 20, 30, 40, 'Fin')
primero, *resto, final = datos

print(primero, resto, final)
print(primero, *resto, final)
# Desempaquetamos los elementos intermedios de datos utilizando *.
```

#### $Slicing$ de Tuplas
Utiliza la notación `tupla[inicio:fin:paso]` para extraer subsecciones de una tupla.

```python
data = (400, 20, 1, 4, 10, 11, 12, 500)

print(data[:4])
print()
print(data[::2])
print()
print(data[::-2])
```

## 3. *Named tuples* (`collections.namedtuple`)

- Permiten asignar **nombres** a cada posición del contenedor.
- Mejoran la legibilidad y claridad del código.
- Al igual que las tuplas estándar, son **inmutables**.
- Poseen el método `_replace(...)` para crear una **nueva copia** modificando campos específicos.

```python
from collections import namedtuple

Info = namedtuple("Info", ["correo", "comuna", "nombre"])
# Creamos la NamedTuple y declaramos su nombre con sus atributos.

tupla = Info("elpisoeslaburo@gmail.com", "Maipú", "Benjamín")
# Inicializamos la tupla

print(tupla.correo)
print(tupla.comuna)
print(tupla.nombre)
print()

tupla_1 = tupla._replace(nombre="Sebastián", correo = "seb.dominguez@uc.cl")
# Creamos una tupla modificada de la anterior, cambiando
# Algunos de sus atributos

print(tupla_1.correo)
print(tupla_1.comuna)
print(tupla_1.nombre)
```
