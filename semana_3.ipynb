# `Semana 3`: Estructuras de datos

- **Estructura de Datos**: Construcción que permite agrupar datos relacionados junto con operaciones para accederlos y modificarlos de manera eficiente.
- **Iterar / Iterable**: Iterar significa *recorrer* una estructura. Un objeto es iterable si *puede ser recorrido*.

### Clasificación de estructuras
1. **Estructuras Secuenciales**:
   - Basadas en un ordenamiento secuencial (depende de cómo se ingresan los elementos).
   - Se pueden recorrer manteniendo el orden que induce la estructura y se puede acceder al $n$-ésimo elemento.
   - *Ejemplos*: `tuple`, `list`, `str`, stacks (pilas) y colas.
2. **Estructuras No Secuenciales**:
   - Almacenan datos sin establecer un orden fijo de acceso.
   - Proveen métodos muy eficientes para la búsqueda de datos.
   - *Ejemplos*: `dict`, `set`, árboles, grafos, heaps.

## 1. Tuple (Tuplas)

Es una estructura de datos **ordenada** e **inmutable** (no se pueden modificar, agregar ni eliminar sus elementos una vez creada).
- Son **heterogéneas**: pueden contener distintas clases o tipos de datos (incluso otras estructuras).
- **Inmutabilidad y Hasheabilidad**:
  - Los objetos inmutables son `hasheables` (permiten calcular un valor `hash` único a partir de su contenido).
  - Si una tupla contiene un elemento mutable (como una lista), deja de ser `hasheable`.
  - Se puede hacer $slycing$ al igual que una lista.

```python
tupla = (1,2,3)
# Creamos una tupla con los elementos 1,2,3
print(f"{tupla}\n")

tupla = 1, True, ("Es 15 de Agosto", "Ayuda", "Retiro ramo?")
# Creamos una tupla sin utilizar paréntesis (son opcionales)
print(f"{tupla}\n")

tupla = ("Grabándote:)",)
# Para crear una tupla con un solo elemento, se debe agregar
# una "," al final, para que no haya error de interpretación.
print(f"{tupla}\n")

tupla = ("Grabándote:)")
print(f"{tupla}\n")
# Sin la coma final

tupla = ()
print(f"{tupla}")
print(type(tupla))
# Vacía
```

#### Verifiquemos la obtención del `hash` e intentemos modificar un elemento de la tupla.

```python
tupla_1 = (1, 2, 3, 4, 5, 6, 7)
tupla_2 = (1, 2, 3, 4, 5, 6, 7)

print(tupla_1)
print(tupla_2)
print()
print(tupla_1.__hash__())
print(tupla_2.__hash__())
# Imprimimos los hash asociados a cada tupla, y como contienen a los mismos elementos,
# tienen el mismo hash.

tupla = (2.4, 1.3, False, "xd")

tupla[1] = True

# Intentamos modificar un elemento de la tupla
```

#### Desempaquetado de Elementos y Operador Asterisco (`*`)
- Las funciones pueden retornar múltiples valores empaquetados en una tupla.
- El operador `*` permite desempaquetar secuencias en variables individuales o al construir nuevas estructuras.

```python
from typing import Tuple

def partes(num : float) -> Tuple[int, float]:
    parte_entera = int(num // 1)
    parte_decimal = num - parte_entera
    return parte_entera, parte_decimal

print(partes(2.45))
print()
# La función partes nos devuelve una tupla con la parte entera y decimal del float ingresado.

datos = ('Inicio', 10, 20, 30, 40, 'Fin')
primero, *resto, final = datos

print(primero, resto, final)
print(primero, *resto, final)
# Desempaquetamos los elementos intermedios de datos utilizando *.
```

## *Named tuples* (`collections.namedtuple`)

- Permiten asignar **nombres** a cada posición del contenedor.
- Mejoran la legibilidad y claridad del código.
- Al igual que las tuplas estándar, son **inmutables**.
- Poseen el método `_replace(...)` para crear una **nueva copia** modificando campos específicos.

```python
from collections import namedtuple

Info = namedtuple("Info", ["correo", "comuna", "nombre"])
# Creamos la NamedTuple y declaramos su nombre con sus atributos.

tupla = Info("elpisoeslaburo@gmail.com", "Maipú", "Benjamín")
# Inicializamos la tupla

print(tupla.correo)
print(tupla.comuna)
print(tupla.nombre)
print()

tupla_1 = tupla._replace(nombre="Sebastián", correo = "seb.dominguez@uc.cl")
# Creamos una tupla modificada de la anterior, cambiando
# Algunos de sus atributos

print(tupla_1.correo)
print(tupla_1.comuna)
print(tupla_1.nombre)
```

## 2. Stacks (Pilas)

Es una estructura de datos de tipo **Last In, First Out (LIFO)**: el último elemento en entrar es el primero en salir.
- **Operaciones clave**:
  - `push`: Agrega un elemento al tope (`stack.append(elem)`).
  - `pop`: Retorna y elimina el elemento del tope (`stack.pop()`).
  - `peek`: Muestra el elemento del tope sin eliminarlo (`stack[-1]`).
  - `is_empty`: Revisa si el stack está vacío (`len(stack) == 0`).
- **Aplicaciones reales**: Navegación con botón "volver" (*back*), reversión de secuencias/textos, deshacer acciones.

## 3. Queues (Colas) y Colas de Doble Extremo

Son una estructura de tipo **First In, First Out (FIFO)**: el primer elemento en entrar es el primero en salir.
- **Operaciones clave**:
  - `enqueue`: Agrega al final (`cola.append(elem)`).
  - `dequeue`: Retorna y extrae el primer elemento (`cola.popleft()`).
- **Eficiencia (`list` vs `deque`)**:
  - Hacer `pop(0)` en una lista tradicional toma $O(N)$ porque desplaza todos los elementos restantes a la izquierda.
  - En un `deque`, extraer o agregar por ambos extremos toma **tiempo constante $O(1)$**.
  - El acceso por índice aleatorio `deque[i]` es $O(N)$, mientras que en `list[i]` es $O(1)$.

  Tanto para **stacks** como para **queues**, usaremos `deques` de **collections**.

```python
from collections import deque

# Crearemos un stack con deque
class PilaTablones:
    num_tablon = 0
    def __init__(self) -> None:
        self.tablones = deque()
    
    # Usamos append para push
    def agregar_tablon(self) -> None:
        self.tablones.append(f"Tablon {self.num_tablon}")
        PilaTablones.num_tablon += 1
    
    # Hacemos peek
    def ver_tablon_tope(self) -> None:
        print(f"El tablón tope es {self.tablones[-1]}")
    
    # Hacemos pop
    def usar_tablon(self) -> None:
        tablon = self.tablones.pop()
        print(f"Se usó el tablon {tablon}!")
        PilaTablones.num_tablon -= 1
        
pila = PilaTablones()

for i in range(5):
    pila.agregar_tablon()

print(pila.tablones)
pila.ver_tablon_tope()

for i in range(2):
    pila.usar_tablon()
    
print(pila.tablones)
```

```python
# Crearemos una queue con deque primeramente
class FilaEstadio:
    def __init__(self) -> None:
        self.fila = deque()
    
    # Usamos append para enqueue
    def agregar_persona(self, nombre: str) -> None:
        self.fila.append(nombre)
        print(f"Agregamos a {nombre} a la fila")
    
    # Hacemos peek
    def ver_primera_persona(self) -> None:
        print(f"La primera persona es {self.fila[0]}")
    
    # Hacemos pop para dequeue
    def atender(self) -> None:
        persona = self.fila.popleft()
        print(f"Se atendió a {persona} y pudo entrar al estadio!")
        
fila_estadio = FilaEstadio()

fila_estadio.agregar_persona("Mati_1")
fila_estadio.agregar_persona("Benja")
fila_estadio.agregar_persona("Martín")
fila_estadio.agregar_persona("Mati_2")

print(fila_estadio.fila)
print()

for i in range(3):
    fila_estadio.atender()
print()

print(fila_estadio.fila)

```

## 4. Dict (Diccionarios)

- Estructura de datos **no secuencial**, **mutable** y basada en el mapeo de **clave-valor** (*key-value*).
- **Eficiencia**: Acceder a un valor por su llave `dict[clave]` toma tiempo constante $O(1)$ independiente del tamaño del diccionario.
- **Requisitos de las Claves**:
  - Deben ser **únicas**.
  - Deben ser **hasheables** (implementar `__hash__` inmutable y `__eq__`). Todos los *built-ins* inmutables (`int`, `str`, `tuple` pura) son *hasheables*.
- **Valores**: No tienen restricciones de tipo.
- **Métodos principales**: `get(key, valor)`, `keys()`, `values()`, `items()`, y diccionarios por comprensión.
> Nota: El segundo argumento de get(key, valor), se refiere al valor de retorno en el caso de que la clave no exista.

```python
nombres = ("Emilio", "Diego", "Paula", "Francisca")

# Creamos el diccionario que almacenará a los
# ramos tomados por las personas de la tupla
# nombres

ramos_tomados = {nombres[i] : i for i in range(4)}

# Usamos dic.items() para poder obtener la tupla
# (clave, valor)

for persona, ramos in ramos_tomados.items():
    print(f"{persona} tomó {ramos} ramos este semestre")

# Agregamos la clave Víctor junto a su valor 2.
ramos_tomados["Victor"] = 2

print(ramos_tomados)
```

## 5. `defaultdict` (`collections.defaultdict`)

- Variante de `dict` que asigna automáticamente un **valor por defecto** cuando se intenta acceder o modificar una llave inexistente.
- Recibe un *callable* (función sin parámetros) como argumento en la inicialización (ej. `int`, `list`, o una función personalizada).
- Evita tener que verificar manualmente si la llave existe antes de operar sobre ella (ideal para conteos de frecuencia o agrupaciones).

```python
from collections import defaultdict
# Inicializamos el defaultdictt e ingresamos
# que sino existe la clave, retorne un int (0)
nombres = defaultdict(int)

nombres[0] = "Eze"
nombres[1] = "Doom"
nombres[2] = "Anakin"

print(nombres[2])
print()
print(nombres[3])

```

## 6. Sets (Conjuntos)

Los *sets* son contenedores **mutables**, **no ordenados** y que **no repiten elementos**. Tienen un comportamiento análogo a los conjuntos matemáticos.
- **Eficiencia**: Revisar si un elemento pertenece o no a un set toma tiempo constante $O(1)$, es decir, su tiempo de operación no depende del tamaño del conjunto.
- **Requisito**: Los *sets* pueden contener cualquier objeto que sea **hasheable** (por ejemplo, los mismos que pueden ser llaves en un diccionario).
- **Operaciones principales**:
  - `add`: Añade un elemento al set. Si se intenta agregar algo que ya estaba, no ocurre nada.
  - `remove` / `discard`: Saca un elemento del set. `remove` lanza un error si el elemento no existe, mientras que `discard` es seguro y no arroja error.
  - **Operaciones de conjuntos**: Soportan la Unión (`|` o `union`), Intersección (`&` o `intersection`), Diferencia (`-` o `difference`) y Diferencia simétrica (`^` o `symmetric_difference`). 

```python
# Creamos un set con set().
# No usamos {} porque se tomaría como
# un diccionario.
conjunto = set()

# También se puede con una lista
conjunto = set([1, 2, 4, 4, 5, True])
# No se añade el True al conjunto,
# puesto que True tiene el valor
# de 1 (al igual que False con 0).

conjunto.add(3)
conjunto.add(3)
# Como es un set, da lo mismo cuántas
# veces intentemos añadir un elemento repetido.
print(conjunto)

conjunto.discard(2)
conjunto.discard(222)
print(conjunto)

conjunto_1 = {4, 5, 6, 7}


print(f"Unión: {conjunto | conjunto_1}")
print(f"Intersección: {conjunto & conjunto_1}")
print(f"Diferencia 1: {conjunto - conjunto_1}")
print(f"Diferencia 2: {conjunto_1 - conjunto}")
print(f"Diferencia simétrica: {conjunto ^ conjunto_1}")
```

```python
# Con >= o <= podemos saber si un conjunto
# es superconjunto o conjunto de otro
# (quitando el igual, es un subconjunto
# o superconjunto propio).
colores_1 = {"verde", "violeta", "azul", "rojo", "amarillo"}
colores_2 = {"amarillo", "azul"}
print(colores_1)
print(colores_2)

# Se podría decir, que donde apunta la flecha
# es el subconjunto.
print(f"¿Es colores_2 subconjunto de colores_1? : {colores_2 <= colores_1}")
print(f"¿Es colores_2 superconjunto de colores_1? : {colores_2 >= colores_1}")
print(f"¿Ambos conjuntos son iguales?: {colores_1 == colores_2}")
```

## 7. Cantidad variable de argumentos (`*args` y `**kwargs`)

> **Nota**: Para añadir parámetros opcionales, basta con añadir un `=` y el valor por defecto.

```python
def paridad(num : int, par = True):
    if par:
        return num % 2 == 0
    else:
        return num % 2 != 0
    
print(paridad(1))            # Es impar: False
print(paridad(1, False))     # Es impar: True
print(paridad(1, par=False)) # Es impar: True
```

Estos conceptos (**args** y **kwargs**) nos permiten especificar una cantidad **variable de argumentos** en la definición de una función.
- **`*args`**: Permite declarar una cantidad arbitraria de **argumentos posicionales**. Al llamarse la función, se reciben esos argumentos y son agrupados internamente en una **tupla** accesible mediante la variable `args`.
- **`**kwargs`**: Permite declarar una cantidad arbitraria de **argumentos por palabra clave** (*keyword arguments*). Durante el llamado de la función, estos argumentos son recibidos y almacenados en un **diccionario** accesible mediante `kwargs`.
- **Desempaquetado**: Adicionalmente, `*` y `**` pueden ser usados al *llamar* a una función. El operador `*` permite desempaquetar una lista o tupla (como vimos anteriormente) y entregar esos valores como argumentos posicionales, y `**` desempaqueta un diccionario para entregarlos como argumentos por palabra clave.

```python
from typing import Union, Tuple
# Se tiene que seguir el siguiente orden para
# definir los argumentos a recibir.
def ejemplo(posicional_1, posicional_2, *args, por_clave = 1, **kwargs):
    print(f"Los posicionales son: {posicional_1} {posicional_2}")
    print(f"Por palabra-clave: {por_clave}")
    print(f"Los desempaquetados: {" ".join(args)}")
    print(f"Los desempaquetados palabra-clave: {kwargs}")
    
ejemplo(10, 20, por_clave="hola", *("1", "xd", "3"), **{"Clave_1": "Valor_1", "Clave_2": "Valor_2"})

# Ejemplo

def sumar_todo(*args : Union[int, float]) -> Union[int, float]:
    suma = 0
    for iter in args:
        suma += iter
    return suma

print(sumar_todo(1, 2, 3, 4, 5))
```
