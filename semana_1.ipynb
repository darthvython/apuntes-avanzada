## Encapsulamiento e interfaces
El encapsulamiento es ocultar métodos y atributos para que solo puedan ser accedidos a ellos como el programador lo quiera.

**Interfaz** : Es el conjunto de métodos y atributos de un objeto que son accedibles por los otros objetos.

**Abstracción** : Es el nivel de detalle de la interfaz.

Para encapsular a un objeto o método, por convención se coloca un **_** antes del atributo o método.
Sin embargo, esto como lo anterior es sólo una **convención**, se podrá seguir utilizando el atributo/método fuera de la clase.
Por esto, para "simular" métodos o atributos privados, se utilizará **__** para el atributo/método.

```python
class Persona:
    def __init__(self,nombre: str, clave: int) -> None:
        self.nombre = nombre
        self.__clave = clave
        
    def decir_clave(self) -> None:
        print(f"{self.nombre}: Mi clave es {self.__clave}")
        
pedro = Persona("Pedro",1234)

pedro.decir_clave()
```

##### Si probamos acceder a la clave de Pedro fuera del método decir_clave:

```python
class Persona:
    def __init__(self,nombre: str, clave: int) -> None:
        self.nombre = nombre
        self.__clave = clave
        
    def decir_clave(self) -> None:
        print(f"{self.nombre}: Mi clave es {self.__clave}")
        
pedro = Persona("Pedro",1234)

print(pedro.__clave)
```

## Archivos y Emulación IO

Podemos emular un archivo con el módulo **io**.

Como **io** crea un archivo en la memoria RAM, no es necesario abrirlo, puesto que ya lo está. De lo anterior, sólo se puede interactuar con métodos de la clase archivo (write(), read(), readlines()).

```python
from io import StringIO, BytesIO

archivo = StringIO()

for i in range(5):
    archivo.write(str(i) + '\n')
    
#Se necesita volver el puntero de lectura a 0
archivo.seek(0)

print(archivo.read())

archivo.close()
```

Además, podemos leer archivos con distintos tipos de codificación:

```python
archivo = open("archivos_1/vacio.txt", "w", encoding = "utf-8")
# Abrimos el archivo con utf-8, que soporta la mayoría de caracteres del mundo.

contenido = "Hola, ¿Cómo estás?" 
archivo.write(contenido)

archivo.close()

# Lo abrimos en lectura en utf-8
archivo = open("archivos_1/vacio.txt", "r", encoding = "utf-8")

print(archivo.read())

archivo.close()

# Lo abrimos en lectura en ascii (caracteres anglo)
# Colocamos el argumento errors para que los caracteres que no se puedan leer se cambien por el símbolo �
archivo = open("archivos_1/vacio.txt", "r", encoding = "ascii", errors = "replace")

print(archivo.read())

archivo.close()
```

Además, con el argumento "w" (write) se sobreescribirá el archivo, mientras que con "a" (append) se agregará lo que escribamos.

Para los argumentos "wb" o "rb", estos abrirán el archivo en modo de escritura o lectura en bytes.
> Nota : Los archivos en bytes se tienen que escribir con `write(b"str")`, siendo str un string cualquiera.

## Context Manager
Un context manager administra la entrada y salida (inicio y cierre) de un recurso para un proceso que queramos hacer, asegurando su limpieza (cerrarlo en la RAM) si ocurre un error.

Un ejemplo, sería crear un context manager que, al ingresar una lista con int's dentro, se encargue de sumarle 1 a todos sus elementos.

```python
class SumarUno:
    def __init__(self, lista):
        self.data = lista
        
    def __enter__(self):
        print("Entrando al Context Manager...")
        return self.data
    # Cuando iniciamos el context manager, se crea una lista
    
    def __exit__(self, exc_type, exc, tb):
        print("Cerrando el context manager...")
        return
    
lista = [1,2,3,4]

with SumarUno(lista) as l:
    for i in range(len(l)):
        l[i] += 1
        
print(lista)
```

Notemos que si la lista no fuera únicamente de int's,

```python
lista = [0, 1, 2, 3, 4, 'hola']

with SumarUno(lista) as l:
    for i in range(len(l)):
        l[i] += 1

print(lista)
```

Aunque nos dé error, el Context Manager nos garantiza que la información no se haya perdido: los elementos antes de llegar "hola" sí fueron modificados en la RAM.

## Anotaciones
Para hacer más legible a nuestro código, para que la IDE nos dé sugerencias, o los errores nos den un Feedback más adecuado, las anotaciones son muy útiles.

```python
from __future__ import annotations
from typing import List, Any, Union
# Del módulo typing y __future__ importamos las anteriores clases.

# Anotamos las variables con anotaciones:
numero : int = 3
string : str = "hola"
lista : list = [1, 2, True, "False"]

# Con typing, podemos indicar que una lista contiene a solo elementos de un tipo:
lista : List[int] = [1, 10, 22]

# Con Any podemos indicar que una variable puede ser cualquier cosa.
variable : Any = 'hola'
variable : Any = 23
variable : Any = False

# Con Union la unión de dos tipos. Por ejemplo, Union[str, int], puede ser un str o un int.
varios : Union[int, float] = 3

# Para métodos o funciones, se puede indicar con -> el retorno

def suma (n1 : int, n2: int) -> int:
    return n1 + n2

# Se puede crear una anotación con objetos de la misma a clase a definir con __future__.
class Contacto:
    def __init__(self : Contacto) -> None:
        self.contactos : List[Contacto] = []
        
```
