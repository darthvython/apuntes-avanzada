## 1. Herencia

Representa una relación de especialización y generalización. Una **subclase** hereda atributos y métodos de una **superclase**, permitiendo reutilizar el código.

- La **especialización** es, que la subclase, normalmente tiende a ser una versión más específica de una superclase.
- La **generalización** es, poder relacionar distintos métodos o atributos que comparten varias clases, para así crear una superclase, que contenga todos aquellos. En resumen, la generalización es lo inverso a la especialización.

- **Sobreescritura (*Overriding*)**: Consiste en volver a definir un método en una subclase con el mismo nombre que tenía en la superclase, con el objetivo de especializar o alterar su comportamiento.
- **Uso de `super()`**: Permite invocar la implementación de un método de la superclase sin necesidad de nombrarla explícitamente (`super().__init__(...)`). Es clave para inicializar atributos heredados correctamente y para crear un código más limpio y escalable ante posibles jerarquías más complejas.
- La **herencia** también se puede hacer con las funciones *built-in* de python.

````python
class Computador:
    def __init__(self, sistema_operativo: str) -> None:
        self.os = sistema_operativo
        
    def prender(self) -> None:
        print(f"Computador Prendido!")
        
    def apagar(self) -> None:
        print(f"Computador Apagado!")
        
class Notebook(Computador):
    # Se heredan automáticamente todos los métodos y atributos de Computador.
    # Sin embargo, si se quiere añadir otros atributos, se debe modificar
    # __init__, por lo que se tiene que invocar nuevamente, y heredar los
    # atributos de Computador.
    
    def __init__(self, marca: str, sistema_operativo : str, screen_size: str) -> None:
        # Le pasamos el atributo sistema_operativo (en orden que lo requiera Computador)
        # y heredamos el atributo
        super().__init__(sistema_operativo)
        
        # Inicializamos los atributos propios de la subclase.
        self.marca = marca
        self.screen_size = screen_size
        
    # Overriding de prender
    def prender(self):
        print(f"[{self.marca}] Computador Prendido!")
        

notebook = Notebook("Asus", "Windows", 14.5)

# Se heredan los métodos de Computador.
notebook.prender()
notebook.apagar()
````

````python
# Asi como también se hereda el atributo de sistema_operativo (os).
print(f"El sistema operativo es {notebook.os}")
print()
print(f"La marca del notebook es {notebook.marca}")
print(f"El tamaño de pantalla es de {notebook.screen_size} pulgadas")
````

## 2. Polimorfismo

Es la propiedad por la que es posible enviar mensajes sintácticamente iguales a objetos de tipos distintos y que cada uno sepa cómo responder.
- **Tipos principales**:
  - ***Overriding***: El mismo de antes; la sobreescritura de métodos es la forma de polimorfismo más natural empleada en Python.
  - ***Overloading***: Es la capacidad de definir un método con el mismo nombre pero distinto número/tipo de parámetros. En Python **no** se soporta el *function overloading* clásico, pero puede simularse empleando argumentos con valores por defecto o cantidad variable de parámetros.
- ***Overloading* de Operadores**: Python sí permite personalizar métodos especiales (`__add__`, `__lt__`, `__eq__`, etc.) para sobrecargar el comportamiento de operadores básicos (como `+`, `<` o `==`) dependiendo de los objetos involucrados.
- **`__repr__` vs `__str__`**: Ambos entregan una representación en texto del objeto. `__str__` está orientado a ser legible por humanos (utilizado por `print`), mientras que `__repr__` busca ser completo y libre de ambigüedades, más orientado a desarrolladores.
- ***Duck typing***: Tipado dinámico: Significa que para invocar una acción sobre un objeto no interesa de qué tipo exacto es, sino únicamente si implementa los métodos requeridos para dicha acción.

````python
from collections import defaultdict
from __future__ import annotations

class Baul:
    numero_de_baul = 0
    
    def __init__(self) -> None:
        self.objetos = defaultdict(int)
        Baul.numero_de_baul += 1
        self.numero = Baul.numero_de_baul
    
    # Los siguientes métodos son los operadores sobrecargados:
    def __getitem__(self, objeto: str) -> int:
        return self.objetos[objeto]
    
    def __setitem__(self, objeto: str, cantidad: int) -> None:
        self.objetos[objeto] = cantidad
    
    # Definimos la suma de baúles
    def __add__(self, otro_baul: Baul) -> Baul:
        nuevo_baul = Baul()
        
        for objeto, cantidad in self.objetos.items():
            nuevo_baul[objeto] += cantidad
            
        for objeto, cantidad in otro_baul.objetos.items():
            nuevo_baul[objeto]  += cantidad
        
        return nuevo_baul
    
    def agregar_objeto(self, objeto: str, cantidad: int) -> None:
        self.objetos[objeto] += cantidad
            
    def __repr__(self) -> str:
        return f"Baul_n{self.numero}"
    
    def items(self) -> list:
        return self.objetos.items()

baul_1 = Baul()
baul_2 = Baul()

baul_1.agregar_objeto("Cuaderno", 3)
baul_2.agregar_objeto("Cuaderno", 4)
baul_1.agregar_objeto("Cargador", 2)

baul_3 = baul_1 + baul_2

print("=========================")
print("‖   Objeto   ‖ Cantidad ‖")
print("=========================")
for objeto, cantidad in baul_3.items():
    print(f"‖ {objeto: ^10} ‖ {cantidad: ^8} ‖")
print("=========================")
````

> Notar que el código anterior, se pudo haber hecho más pequeño, heredando la clase **defaultdict**.

````python
class Baul(defaultdict):
    numero_de_baul = 0
    
    def __init__(self) -> None:
        super().__init__(int)
        Baul.numero_de_baul += 1
        self.numero = Baul.numero_de_baul
        
    # Definimos la suma de baúles
    def __add__(self, otro_baul: Baul) -> Baul:
        nuevo_baul = Baul()
        
        for objeto, cantidad in self.items():
            nuevo_baul[objeto] += cantidad
            
        for objeto, cantidad in otro_baul.items():
            nuevo_baul[objeto]  += cantidad
        
        return nuevo_baul
    
    def agregar_objeto(self, objeto: str, cantidad: int) -> None:
        self[objeto] += cantidad
           
    def __repr__(self) -> str:
            return f"Baul_n{self.numero}"
        
baul_1 = Baul()
baul_2 = Baul()

baul_1.agregar_objeto("Cuaderno", 3)
baul_2.agregar_objeto("Cuaderno", 4)
baul_1.agregar_objeto("Cargador", 2)

baul_3 = baul_1 + baul_2

print("=========================")
print("‖   Objeto   ‖ Cantidad ‖")
print("=========================")
for objeto, cantidad in baul_3.items():
    print(f"‖ {objeto: ^10} ‖ {cantidad: ^8} ‖")
print("=========================")
````

## 3. Multiherencia

Se presenta cuando una subclase hereda atributos y comportamientos de **más de una superclase** de manera simultánea.
- **Problema del diamante**: Ocurre en jerarquías donde una clase inferior llega a un ancestro común a través de múltiples caminos. Esto genera el riesgo de invocar múltiples veces funciones que deberían ocurrir solo una (como la inicialización o contadores).
- **MRO (Method Resolution Order)**: Es el orden que sigue Python internamente (usando el algoritmo *C3 Linearization*) para decidir qué método ejecutar en casos de herencias múltiples o conflictos de nombres. Se puede visualizar este orden en cualquier clase llamando al método predefinido `__mro__`.

````python
class Android:
    def __init__(self, version_os : str):
        self.os = "Android"
        self.version_os = version_os
        
class Tablet:
    def __init__(self, precio : float, screen_size: float, almacenamiento: float) -> None:
        self.precio = precio
        self.screen_size = screen_size
        self.almacenamiento = almacenamiento
        
class SamsungTab(Android, Tablet):
    def __init__(self,  modelo : str, version_os, precio : float, screen_size : float, 
                 almacenamiento : float):
        Android.__init__(self, version_os)
        Tablet.__init__(self, precio, screen_size, almacenamiento)
        
        self.modelo = modelo
        
tablet = SamsungTab("Tab S11 Ultra", "Android 17", 1700, 14.6, 512)

print(f"Modelo : {tablet.modelo}")
print(f"Sistema Operativo : {tablet.os}")
print(f"Versión Sistema Operativo: {tablet.version_os}")
print(f"Precio : {tablet.precio} dólares")
print(f"Tamaño de pantalla : {tablet.screen_size} pulgadas")
print(f"Almacenamiento : {tablet.almacenamiento}gb")

````

#### Solucionando los conflictos de Multiherencia

**El uso avanzado de `super()`** es para evitar llamadas duplicadas a métodos de la clase base (el centro del problema del diamante), la mejor práctica es emplear siempre `super()` en lugar de llamar directamente a `Superclase.__init__`. Esto garantiza que Python respete la cadena dictada por el MRO.
- El **uso de `*args` y `**kwargs`** en multiherencias, se suele emplear cuando las superclases esperan parámetros distintos. La forma óptima de resolver esto es enviar todos los parámetros agrupados mediante diccionarios en `**kwargs`. Así:
  1. Cada inicializador absorbe del `**kwargs` los parámetros con nombre que le corresponden.
  2. Envía el resto de los argumentos empaquetados hacia la siguiente clase de la lista MRO haciendo un llamado a `super().__init__(**kwargs)`.
- **Alternativas**: Ocasionalmente, si el sistema es muy rígido o complejo, existe la opción (aunque no siempre recomendada por ser menos dinámica) de obviar `super()` y hacer las llamadas a cada clase directamente en la subclase inferior. Toda solución depende fuertemente de la arquitectura y del mismo programador.

````python
class Animal:
    def __init__(self, nombre, **kwargs):
        self.nombre = nombre
    
class Mamifero(Animal):
    def __init__(self, temperatura_sangre : float, **kwargs):
        super().__init__(**kwargs)
        self.temperatura_sangre = temperatura_sangre

class AnimalAcuatico(Animal):
    def __init__(self, profundidad_maxima : float, **kwargs):
        super().__init__(**kwargs)
        self.profundidad_maxima = profundidad_maxima
        
class Ballena(Mamifero, AnimalAcuatico):
    def __init__(self, largo : float, **kwargs):
        super().__init__(**kwargs)
        self.largo = largo
        
print(Ballena.__mro__)

# Se ve que el camino del MRO es:
# Ballena -> Mamifero -> AnimalAcuatico -> Animal -> Object
````

````python
ballena = Ballena(15, nombre = "Ballena Jorobada", temperatura_sangre = 37, profundidad_maxima = 300)

# Este objeto de construye de la siguiente manera:

# -- Ballena (toma largo)
# -- kwargs --> Mamifero (quita de kwargs temperatura_sangre)
# -- kwargs --> AnimalAcuatico (quita de kwargs profundidad_maxima)
# -- kwargs --> Animal (quita de kwargs nombre)
# -- kwargs (vacío) --> object (SuperClase de todas las clases)

# Ahora es el flujo de regreso:
# object -> Crea el objeto en sí.
# Animal -> self.nombre = nombre
# AnimalAcuatico -> self.profundidad_maxima = profundidad_maxima
# Mamifero -> self.temperatura_sangre = temperatura_sangre

# Se termina el flujo de super()
# Se sigue en Ballena con:
# self.largo = largo

print(ballena.nombre)
print(ballena.temperatura_sangre)
print(ballena.profundidad_maxima)
print(ballena.largo)
````

## 4. Decoradores

Un `decorador` es una función (*callable/invocable*) que recibe otra función como input y retorna una versión modificada de ella. Permiten agregar o modificar el comportamiento de una función (o clase) **sin cambiar su código original**.
- Los decoradores son posibles porque en Python las funciones son **objetos de primera clase**, es decir:
    - Se pueden **asignar a variables** (sin paréntesis para no ejecutarlas).
    - Se pueden **pasar como argumento** a otra función (o ser retornables).
    - Se pueden almacenar en `estructuras de datos`.
    - Se pueden acceder a sus atributos.

````python
from typing import Callable
# Callable es para describir un argumento 
# que es invocable.

def dormir(nombre : str) -> str:
    return f"[{nombre}] Se durmió"

# Le asignamos el nombre de descansar a dormir
# sin paréntesis para referirnos a la función
# en sí y no invocarla.
descansar = dormir

print(descansar("Juan Pablo"))

# Podemos crear funciones que reciban argumentos
# como funciones.
def despertar_jp(funcion : Callable) -> str:
    return funcion("Juan Pablo") + " y se despertó a los 5 minutos"

print(despertar_jp(descansar))
````

### Sintaxis de los decoradores

La sintaxis `@nombre_decorador` sobre una función es una abreviación de:

```python
def mi_funcion():
    pass

mi_funcion = decorador(mi_funcion)
```

Es decir, `decorador` recibe la función original, la envuelve dentro de una nueva función (definida dentro de él) y retorna esa nueva función, que reemplaza a la original.

````python
def decorador(func : Callable) -> Callable:
    
    # Definimos la función a retornar
    def envoltorio(num : float) -> str:
        # Se debe recibir los parámetros
        # de la original, y retornar o realizar
        # los cambios que queremos moldear
        # a ésta.
        
        return f"El resultado es {func(num)}"
    
    # Retornarmos la función
    return envoltorio 

@decorador 
def cuadrado(num : float) -> float:
    return num*num


print(cuadrado(4))
````

### `@property`

En Python todos los atributos son públicos; el *underscore* (`_atributo`) es solo una convención para indicar que un atributo debiese tratarse como privado. Para proteger o controlar el acceso/escritura de un atributo se usa `property`, que permite definir un *getter*, un *setter* (y un *deleter*) para un atributo, dejándolo con apariencia de atributo normal aunque por dentro ejecute lógica (por ejemplo, validaciones).

Se puede definir de dos formas equivalentes:
- Usando la función *built-in* `property(fget, fset, fdel)` directamente.
- Usando la sintaxis de decoradores: `@property` para el *getter* y `@nombre.setter` para el *setter*.

Utilizaremos esta última.

````python
class Cancion:
    def __init__(self, nombre : str, artista : str, genero : str, duracion : int):
        self.nombre = nombre
        self.artista = artista
        self.genero = genero
        self._duracion = duracion
        
    @property
    def duracion(self) -> str:
        return f"La duración es de {self._duracion // 60:0>2}:{self._duracion % 60:0>2}."
    
    # Colocando @property, y sin agregar un setter,
    # el atributo pasa a ser de solo lectura.
    # Como queremos que sea modificable, entonces
    # agregamos un setter.
    
    @duracion.setter
    def duracion(self, nueva_duracion) -> None:
        if type(nueva_duracion) not in (int, float):
            print(f"Ingresa un int.")
        
        elif nueva_duracion < 0:
            print(f"La duración no puede ser negativa.")
            
        else:
            self._duracion = nueva_duracion
            print(f"Nueva duración asignada : {self._duracion // 60:0>2}:{self._duracion % 60:0>2}!")
        
cancion = Cancion("DEVIL IS A LIE", "Tommy Richman", "Hip-Hop", 131)

print(cancion.duracion)

cancion.duracion = -1

cancion.duracion = "Es demasiada materiaaaaaaaaaa"

cancion.duracion = 150
````

### Métodos estáticos: `@staticmethod`

Un **método estático** es un método de una clase que **no necesita `self`** (no accede ni modifica el estado de una instancia particular). Se declara con el decorador `@staticmethod` y se puede invocar directamente desde la clase, sin necesidad de crear una instancia (`Clase.metodo(...)`).

> **Nota**: Clases solo con `@staticmethod`.
Si una clase solo tiene métodos estáticos y no usa estado (`self`), en Python es absurdo (herencia de Java/C#). Es mejor usar un módulo (`archivo.py`) con funciones sueltas.

## 5. Clases **Abstractas**

Una `clase abstracta` es creada con el objetivo que **no debe instanciarse directamente**; su propósito es servir de molde para que otras clases (subclases) hereden de ella e implementen los detalles específicos. Puede declarar **métodos abstractos**: métodos sin una implementación completa (o obligatoria) que **toda subclase debe sobreescribir**.

Python no tiene clases abstractas de forma nativa, pero el módulo `abc` (*Abstract Base Class*) permite crearlas:
- La clase hereda de `ABC`.
- Los métodos que se quieren obligar a implementar se marcan con `@abstractmethod`.
- Si se intenta instanciar directamente una clase abstracta (o una subclase que no implementa **todos** sus métodos abstractos), Python lanza un error (`TypeError`).

Una clase abstracta **no está obligada a contener solo métodos abstractos**: puede combinar métodos concretos (con lógica ya implementada, heredable tal cual) junto a métodos abstractos (que cada subclase debe redefinir). Una subclase incluso puede reutilizar la implementación "base" de un método abstracto llamando a `super().metodo()` dentro de su propia redefinición.

> Nota: En este ámbito, algo abstracto es más sinónimo a algo incompleto, que a algo abstracto como tal.

````python
from abc import ABC, abstractmethod, ABCMeta

# Heredamos de ABC
class Vivienda(ABC):
    def __init__(self) -> None:
        self._habitantes = []
        
    # Utilizamos el decorador
    # para indicar que es un método abstracto
    @abstractmethod
    def agregar_habitante(self, nuevo_habitante : str) -> None:
        self._habitantes.append(nuevo_habitante)
        print(f"Habitante añadido : {nuevo_habitante}!")
    
    @abstractmethod
    def quitar_habitante(self, habitante : str) -> None:
        if habitante in self._habitantes:
            self._habitantes.remove(habitante)
            print(f"Habitante purgado :( : {habitante}")
            
        else:
            print("No existe el habitante.")
            
    @property
    def habitantes(self):
        return f"Los habitantes son: {', '.join(self._habitantes)}."

# Intentemos instanciar un objeto de Vivienda

vivienda = Vivienda()
````

````python
# Heredamos de Vivienda
class Casa(Vivienda):
    # Reimplementamos el método con algunos
    # ajustes.
    def agregar_habitante(self, nuevo_habitante : str) -> None:
        if len(self._habitantes) < 6:
            
            self._habitantes.append(nuevo_habitante)
            print(f"Habitante añadido : {nuevo_habitante}!")
            
        else:
            print("Son muchas personas :(")
    
    def quitar_habitante(self, habitante : str) -> None:
        super().quitar_habitante(habitante)
    
class Departamento(Vivienda):
    
    def agregar_habitante(self, nuevo_habitante : str) -> None:
        if len(self._habitantes) < 3:
                    
            self._habitantes.append(nuevo_habitante)
            print(f"Habitante añadido : {nuevo_habitante}!")
                    
        else:
            print("Son muchas personas :(")
            
    def quitar_habitante(self, habitante : str) -> None:
        super().quitar_habitante(habitante)
        
# En ambas subclases heredamos los métodos abstractos.
# Si esto último no se hace, se arrojará un error.

casa = Casa()
departamento = Departamento()

casa.agregar_habitante("Germán")
casa.agregar_habitante("Fernanda")
casa.agregar_habitante("Luis")
casa.agregar_habitante("Paulina")
casa.agregar_habitante("Verónica")
casa.agregar_habitante("Benja")
casa.agregar_habitante("José")
casa.quitar_habitante("Benja")

print()

departamento.agregar_habitante("Nathan")
departamento.agregar_habitante("Enzo")
departamento.agregar_habitante("Tom")
departamento.agregar_habitante("Bastián")
````

## 6. Diagrama de Clases

El `diagrama de clases` (UML) permite visualizar las clases de un sistema, sus atributos, métodos y las relaciones entre ellas.

Se dibuja como un rectángulo dividido en 3 niveles:
1. **Nombre** de la clase.
2. **Atributos** (o *properties*), con su tipo.
3. **Métodos**, con sus parámetros y el tipo de retorno.

Una clase puede no tener atributos y/o no tener métodos.

Algunas de sus relaciones son:

- **Contención** (rombo ⧫/◊): una clase tiene a otra como atributo.
  - **Composición** ⧫: el objeto contenido **depende** del tiempo de vida del contenedor (si se elimina el contenedor, se elimina lo contenido).
  - **Agregación**  ◊: el objeto contenido es **independiente** del contenedor (puede seguir existiendo si el contenedor desaparece).
- **Herencia** (flecha ᐅ que apunta a la superclase): relación de especialización/generalización; la subclase hereda atributos y métodos de la superclase (no se vuelven a escribir en el diagrama de la subclase).

## 7. Metaclases

Todo en Python es un objeto, y todo objeto es instancia de una clase (su *tipo*). Esto también aplica a las clases mismas: una clase es instancia de otra clase, llamada **metaclase**. En otras palabras, una metaclase es una clase cuyas instancias son, a su vez, clases.
Su **objetivo** es controlar la creación de clases.

`type(objeto)` retorna la clase (tipo) de un objeto. Si el objeto entregado es a su vez una clase, `type()` retorna su metaclase. Por defecto, la metaclase de toda clase en Python es `type` (a menos que se especifique otra explícitamente).

Para crearlas:
- Se **asignan** a una clase con el *keyword argument* `metaclass=MiMetaclase` en su definición.
- Se **crean** heredando de `type`, y su propósito es controlar cómo se **crean** e **instancian** las clases que la usan.

Para lograr ese control se implementan (opcionalmente) tres métodos:

- **`__new__(cls, nombre, clases_base, diccionario)`**: controla la **creación de la clase** en sí. Se ejecuta al definirse la clase (no al instanciarla). Recibe la metaclase (`cls`), el nombre de la clase, sus clases base y un diccionario con su contenido; puede usarse, por ejemplo, para cambiar el nombre, forzar herencia o modificar la clase antes de crearla.
- **`__init__(self, nombre, clases_base, diccionario)`**: controla lo que ocurre **después de creada la clase** (`self` aquí representa la clase recién creada). Es útil para agregarle atributos o métodos adicionales, pero no para modificar su creación misma.
- **`__call__(cls, *args, **kwargs)`**: controla qué ocurre cuando se **instancia** la clase (es decir, cuando se hace `NombreClase(...)`).

````python
class Prueba(type):
    def __new__(cls, nombre, clases_heredadas, diccionario):
        
        # cls : La propia metaclase (Prueba)
        # nombre : En este caso, es Hola (por la clase de abajo)
        # clases_heredadas : Las clases que pudiese heredar Hola.
        # diccionario:
        print(f"Se está creando una clase de la metaclase {cls.__name__}")
        clase = super().__new__(cls, nombre, clases_heredadas, diccionario)
        
        return clase
    
    
class Hola(metaclass=Prueba):
    pass

# Imprimimos la metaclase de Hola
print(type(Hola))
````

````python
class Prueba(type):
    def __new__(cls, nombre, clases_heredadas, diccionario):
        print(f"Se está creando una clase de la metaclase {cls.__name__}")
        clase = super().__new__(cls, nombre, clases_heredadas, diccionario)
        
        return clase
    
    
class Hola(Vivienda, metaclass=Prueba):
    pass
````

Este error ocurre, porque **se está heredando de Vivienda y de la metaclase Prueba**. 

Si miramos arriba, Vivienda está heredando ABC, y ésta por sí misma hereda de `ABCMeta`, otra metaclase. Por lo que existe un conflicto de metaclases.

Para solucionar esto:

````python
# Heredamos de ABCMeta
class Prueba(ABCMeta):
    def __new__(cls, nombre, clases_heredadas, diccionario):
        
        print(f"Se está creando una clase de la metaclase {cls.__name__}")
        
        # Igualmente queremos prohibir que Hola herede de Vivienda,
        # entonces:
        clases = list(clases_heredadas)
        
        if Vivienda in clases:
            print("Se intentó heredar de Vivienda, pero eso está prohibido")
            clases.remove(Vivienda)
        
        clases = tuple(clases)
        
        clase = super().__new__(cls, nombre, clases, diccionario)
        
        return clase
    
    
class Hola(Vivienda, metaclass=Prueba):
    pass

print(type(Hola))
````

````python
class Prueba(ABCMeta):
    def __new__(cls, nombre, clases_heredadas, diccionario):
        
        print(f"Se está creando una clase de la metaclase {cls.__name__}")
        
        clases = list(clases_heredadas)
        
        if Vivienda in clases:
            print("Se intentó heredar de Vivienda, pero eso está prohibido")
            clases.remove(Vivienda)
        
        clases = tuple(clases)
        
        clase = super().__new__(cls, nombre, clases, diccionario)
        
        return clase
    
    # Queremos agregarle atributos
    def __init__(cls, nombre, clases, diccionario):
        # Inicializamos la clase
        super().__init__(nombre, clases, diccionario)

        # __init__ de la subclase, obtenida de diccionario.
        init_original = diccionario.get('__init__')
        
        # Definimos una función que actuará como el __init__ de las futuras instancias
        def mi_init_de_instancia(self, *args, **kwargs):
            self.prueba = "Este es un atributo de prueba"
            self.prueba_2 = 23

            # Si se escribió __init__ en la subclase
            # colocamos esto para que no se borren los
            # atributos asignados.
            if init_original:
                init_original(self, *args, **kwargs)
        
        cls.__init__ = mi_init_de_instancia
        
    # Se ejecuta cada vez que se instancia un objeto de la subclase.
    def __call__(cls, *args, **kwargs):
        print(f"Se creó una nueva instancia de {cls.__name__}")
        
        return super().__call__(*args, **kwargs)
    
class Hola(Vivienda, metaclass=Prueba):
    variable_de_clase = 100
    
    def __init__(self):
        self.atributo = "Hola xd"

    def saludar(self):
        print("¡Hola!")

print()

hola = Hola()
print(hola.prueba)
print(hola.prueba_2)
print(hola.atributo)
````
