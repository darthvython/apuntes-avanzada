## 1. Excepciones

Una **excepción** es una condición anómala o inesperada que ocurre durante la ejecución de un programa y que interrumpe su ejecución esperada. Todas las excepciones son clases que heredan de `Exception`.

#### 1.1 Tipos comunes de excepciones

- **`SyntaxError`**: el código está mal escrito y viola las reglas sintácticas del lenguaje. Se detecta **antes** de ejecutar el programa (al leerlo).
- **`IndentationError`**: hereda de `SyntaxError`; ocurre cuando un bloque de código (`if`, `for`, `def`, etc.) no respeta la indentación.
- **`NameError`**: se intenta usar un nombre (variable, función, clase) que no ha sido definido en anteriormente.
- **`ZeroDivisionError`**: ocurre al dividir por cero. A diferencia de los errores anteriores, solo puede detectarse en tiempo de **ejecución**, dependiendo de los valores usados.
- **`IndexError`**: se accede a un índice fuera del rango válido de una secuencia (lista, tupla, etc.).
- **`KeyError`**: se accede a una llave inexistente en un diccionario (equivalente a `IndexError` pero para *mappings*).
- **`AttributeError`**: se usa un método o atributo que no existe en una clase o tipo de dato.
- **`TypeError`**: se opera con un argumento de un tipo no soportado para esa operación (ej. sumar un `int` con una `list`, o "llamar" con `()` un objeto que no implementa `__call__`).
- **`ValueError`**: el tipo del argumento es correcto, pero su **valor** no es apropiado para la operación (ej. `"".split("")`, o `lista.remove(x)` con un `x` que no está en la lista).

## 2. Levantando y manejando excepciones

Podemos generar una excepción manualmente en el momento que queramos usando `raise`, creando una instancia de un tipo de excepción con un mensaje explicativo (ej. `raise TypeError("mensaje")`). Al ejecutarse un `raise`, la ejecución se interrumpe inmediatamente: ninguna línea posterior en esa función (ni en las funciones que la llamaron) se ejecuta, a menos que la excepción sea capturada antes de llegar al sistema operativo.

```python
def combinar_listas(lista_1 : list, lista_2 : list):
    if (type(lista_1), type(lista_2)) != (list, list):
        # Creamos una instancia de TypeError
        raise TypeError(f"Ingresa dos listas como argumento, no ({lista_1}, {lista_2})")
    
    else:
        return sorted(lista_1 + lista_2)
    
lista_1 = [1, 3, 5, 7]
lista_2 = [2, 4, 6, 8]

print(combinar_listas(lista_1, lista_2))
print(combinar_listas(lista_1, "Mañana es la I1 de Cálculo ayuda no he estudiado"))
```

### Manejo de excepciones
El bloque `try` define un bloque de código; si se levanta una excepción dentro de él, el flujo salta al `except` correspondiente al tipo de la excepción, y el programa **continúa** después del bloque (no se interrumpe todo el programa).

- Se pueden definir **varios bloques `except`** para dar un tratamiento distinto según el tipo de excepción.
- Se puede capturar **más de un tipo a la vez** en un mismo `except` usando una tupla: `except (TipoA, TipoB) as error:`.
- Solo se pueden atrapar excepciones que ocurren **en tiempo de ejecución**; `SyntaxError`/`IndentationError` no se pueden atrapar en el mismo archivo donde ocurren.
- El bloque **`else`** se ejecuta **solo si no se lanzó ninguna excepción** en el `try`.
- El bloque **`finally`** se ejecuta **siempre**, haya ocurrido o no una excepción.

```python
# Usaremos la función anteriormente definida
try: 
    combinacion = combinar_listas(lista_1, lista_2)
    
    # error representa el string que se imprime
    # al levantarse la excepción
except TypeError as error:
    print(error)
    print("Vuelve a ingresar los valores nuevamente")
    
else:
    print("Se pudo combinar sin inconvenientes...")
    print(combinacion)
    
finally:
    print("Siempre utiliza las excepciones para estos casos")
```

```python
try: 
    combinacion = combinar_listas(lista_1, 1)
    
    # error representa el string que se imprime
    # al levantarse la excepción
except TypeError as error:
    print(error)
    print("Vuelve a ingresar los valores nuevamente")
    
else:
    print("Se pudo combinar sin inconvenientes...")
    print(combinacion)
    
finally:
    print("Siempre utiliza las excepciones para estos casos")
```

## 3. Iterables y Generadores

- Un **iterable** es un objeto que contiene una colección de elementos y puede ser recorrido. Implementa `__iter__()`, el cual retorna un **iterador** (o bien implementa `__getitem__()` como alternativa). Un iterable se puede recorrer **múltiples veces**.
- Un **iterador** es el objeto retornado por `__iter__()`, encargado de entregar los elementos uno a uno mediante `__next__()`. Cuando ya no quedan elementos, debe levantar `StopIteration`. Un iterador implementa también `__iter__()`, que simplemente retorna `self`. Por eso **un iterador es también un iterable**, pero solo se puede recorrer **una vez** (mantiene su propio progreso/memoria de dónde va).
- `iter(objeto)` equivale a llamar `objeto.__iter__()`, y `next(objeto)` equivale a `objeto.__next__()`.

```python
from typing import Self, Any
# Intentaremos crear una "copia" de range.

# Creamos el iterador
class RangeIterador:
    def __init__(self, inicio : int, fin : int):
        self.actual = inicio
        self.fin = fin
        
    def __iter__(self) -> Self:
        return self
        # Retornarmos self, en el caso
        # que se requiera utilizar al 
        # mismo iterador en vez del iterable
        
    def __next__(self) -> int:
        
        if self.actual < self.fin:
            # Avanzamos una posición en el rango
            actual = self.actual
            self.actual += 1
            return actual
        
        # Cuando self.actual >= self.fin
        # entonces queremos que deje de avanzar
        # el bucle for
        
        else:
            raise StopIteration()
        
class RangeIterable:
    def __init__(self, inicio : int, fin : int):
        self.inicio = inicio
        self.fin = fin
        
    def __iter__(self):
        return RangeIterador(self.inicio, self.fin)
    
    # Para visualizar el rango
    def __repr__(self):
        return "(" + ", ".join([str(i) for i in RangeIterable(self.inicio, self.fin)]) + ")"
    
for num in RangeIterable(0,5):
    print(num)
    
print(RangeIterable(0, 5))
```

```python
# El flujo de lo anterior funciona así

# RangeIterable(1,5) crea un objeto iterable con inicio = 1, fin = 5
# Python llama a __iter__() --> Crea un RangeIterador(1,5) nuevo

# Python llama repetidamente a __next__() en ese iterador:
# Primera llamada: retorna 1, actual pasa a 2
# Segunda llamada: retorna 2, actual pasa a 3
# Tercera llamada: retorna 3, actual pasa a 4
# Cuarta llamada: retorna 4, actual pasa a 5
# Quinta llamada: actual (5) ya no es < fin (5) --> lanza StopIteration()

# El for termina y el programa continúa
```

### Generadores

Los **generadores** son un caso especial de iterador: permiten recorrer una secuencia sin almacenar todos sus elementos en memoria (los va generando uno a uno, "sobre la marcha"). Igual que un iterador normal, un generador solo se puede recorrer **una vez**.

- La forma más simple de crear uno es con una **expresión generadora** (casi idéntica a las *list comprehension*), usando paréntesis: `(expresion for elemento in iterable)`.

```python
# Creamos un generador
generador = (i for i in range(10))

for i in generador:
    print(i, end=" ")
    
print("\nPrimera y última iteración sobre generador\n")

for i in generador:
    print(i, end=" ")

print("Ups... ya no se puede utilizar el generador...")
```

## 4. Listas Ligadas

### `Nodo`

Un **nodo** es la unidad básica de muchas estructuras de datos: contiene un **dato** y mantiene referencias a otros nodos ("nodos vecinos").

### `Lista ligada`

Una **lista ligada** almacena nodos en orden secuencial, donde cada nodo tiene una referencia a su **sucesor** (`siguiente`). El primer nodo es la **cabeza** (*head*) y el último (sin sucesor) es la **cola** (*tail*). Para acceder a cualquier nodo que no sea la cabeza, hay que recorrer la lista siguiendo las referencias desde la cabeza.

Las operaciones básicas de una lista ligada son:
- **`agregar(valor)`**: agrega un nuevo nodo al final de la lista (después de la cola).
- **`obtener(posicion)`**: recorre la lista desde la cabeza hasta llegar al nodo en la posición pedida y retorna su valor.
- **`insertar(valor, posicion)`**: inserta un nuevo nodo en una posición específica, redirigiendo las referencias de los nodos vecinos.

```python
from __future__ import annotations

class Nodo:
    # Valor puede ser cualquier cosa
    def __init__(self, valor : Any):
        self.valor = valor
        self.siguiente = None
    
    def __repr__(self):
        return f"[{self.valor}]"
    
# Antes de definir ListaLigada, definiremos un Iterador para ésta:

class IteradorLista:
    def __init__(self, lista_ligada : ListaLigada):
        self.actual = lista_ligada.cabeza
        
    def __iter__(self):
        return self
    
    def __next__(self):
        # El nodo se recorre de manera similar
        if self.actual != None:
            actual = self.actual
            self.actual = self.actual.siguiente
            return actual
        
        else:
            raise StopIteration()

# Ahora sí, definamos ListaLigada
class ListaLigada():
    def __init__(self):
        self.cabeza = None
        self.cola = None
        self.largo = 0
    
    # Para iterar
    def __iter__(self) -> IteradorLista:
        return IteradorLista(self)
        
    
    def agregar(self, valor: Any):
        
        nueva_cola = Nodo(valor)
        
        # Si la lista es vacía
        if not self.cabeza:
            self.cabeza = nueva_cola
            self.cola = self.cabeza

        else:
            self.cola.siguiente = nueva_cola
            self.cola = nueva_cola
            
        print(f"Se agregó el nodo {nueva_cola} al final de la lista!")
        self.largo += 1
        
    def obtener(self, pos : int):
        if not 0 <= pos <= self.largo - 1:
            raise ValueError(f"Ingresa una posición dentro del rango (0, {self.largo - 1})")
        
        # Vamos iterando hasta la posición
        # -> zip es para que pare en la posición deseada
        for nodo, _ in zip(self, range(pos + 1)):
            pass
        
        # Ahora se está en el nodo de la posición que queremos
        return nodo
        
    def insertar(self, valor : Any, pos : int):
        if not 0 <= pos <= self.largo:
            raise ValueError(f"Ingresa una posición dentro del rango (0, {self.largo})")
        
        nuevo_nodo = Nodo(valor)
        
        if not self.cabeza:
            self.cabeza = nuevo_nodo
            self.cola = self.cabeza
            print(f"Se agregó el nodo {nuevo_nodo} en la posición 0!")
            self.largo += 1
            return
        
        # Insertar en la cabeza
        if pos == 0:
            nuevo_nodo.siguiente = self.cabeza
            self.cabeza = nuevo_nodo
            print(f"Se agregó el nodo {nuevo_nodo} en la posición 0!")
            self.largo += 1
            return
        # Insertar en la cola
        elif pos == self.largo:
            self.cola.siguiente = nuevo_nodo
            self.cola = nuevo_nodo
    
        else:
            
            for nodo_actual, num in zip(self, range(pos + 1)):
                # Obtenemos el nodo anterior
                # aunque el bucle se termina hasta
                # encontrar el que queremos
                if num == pos - 1:
                    nodo_anterior = nodo_actual
        
            nuevo_nodo.siguiente = nodo_actual
            nodo_anterior.siguiente = nuevo_nodo
        
        print(f"Se agregó el nodo {nuevo_nodo} en la posición {pos}!")
        self.largo += 1
        
    def __repr__(self):
        return " -> ".join([str(nodo) for nodo in self])
    
lista_ligada = ListaLigada()

lista_ligada.agregar("1")
lista_ligada.agregar("5")
lista_ligada.agregar("10")
lista_ligada.agregar("111")

print(f"\n{lista_ligada}\n")
lista_ligada.insertar("4", 2)

print(f"\n{lista_ligada}\n")
```

## 5. Excepciones personalizadas

### Jerarquía de excepciones

Todas las excepciones heredan de `BaseException`. De ella se desprenden tres grandes tipos: `SystemExit`, `KeyboardInterrupt` y `Exception`. Todas las excepciones producidas por errores durante la ejecución normal de un programa son subclases de `Exception`.

Capturar `Exception` directamente (`except Exception`) atrapa **cualquier** error que sea instancia de una subclase suya, dando un tratamiento genérico. En general, es más recomendable manejar excepciones de forma **selectiva**, capturando el tipo específico esperado (`TypeError`, `ValueError`, etc.), ya que así se distingue qué error ocurrió realmente y se puede reaccionar de forma adecuada a cada uno.

### Crear excepciones personalizadas

Podemos crear nuestros propios tipos de excepción heredando de `Exception`. Si no se sobreescribe nada, la subclase hereda todo el comportamiento de `Exception` sin cambios. También se puede sobreescribir `__init__` para, por ejemplo, construir el mensaje de error a partir de los parámetros recibidos, o para agregar **atributos y métodos propios** que permitan recuperar información adicional sobre el error ocurrido (por ejemplo, guardar los valores que causaron el problema, o un método que calcule algo relacionado al error).

```python
class DatosException(Exception):
    def __init__(self, list_1 : list, list_2 : list):
        # Ingresamos el mensaje a mostrar
        super().__init__(f"Ingresa dos listas como argumento, no ({list_1}, {list_2})")
        
        
# Tomemos la función anteriormente definida,
# ahora con la nueva excepción.
def combinar_listas(lista_1 : list, lista_2 : list):
    if (type(lista_1), type(lista_2)) != (list, list):
        # Creamos una instancia de DatosException
        raise DatosException(lista_1, lista_2)
    else:
        return sorted(lista_1 + lista_2)


lista_1 = [1, 3, 5, 7]
lista_2 = [2, 4, 6, 8]

print(combinar_listas(lista_1, lista_2))
print(combinar_listas(lista_1, "Borren el IMT2210"))
```
