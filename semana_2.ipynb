## Paths y Manejo de Rutas

Un **path** (o ruta) es la dirección que indica la ubicación de un archivo o directorio dentro del árbol del sistema de archivos.

### 1. Tipos de Paths
**Path Absoluto**: Comienza desde la raíz del sistema de archivos (`/` en Linux/macOS o `C:\` en Windows). Es único pero **no es portable**, ya que depende de la estructura exacta del computador donde se ejecuta.

**Path Relativo**: Se calcula desde el directorio de trabajo actual. Es **altamente recomendado** para que el código funcione en cualquier equipo sin romperse.

### 2. Componentes de una Ruta
Una ruta se divide en:
- `dirname`: Nombre del **directorio** donde se encuentra el archivo.
- `basename/filename`: Nombre propio del archivo, incluyendo la extensión.
- `extensión`: Secuencia tras el último punto (`.txt`, `.jpg`). Le indica al sistema cómo abrir el archivo (convención), pero **no determina su contenido real**.

````python
import os
# Importamos el módulo os

ruta = "archivos_2/ejemplo.txt"

print(os.path.dirname(ruta))
# directorio

print()

print(os.path.basename(ruta))
# nombre del archivo
````

### 3. Portabilidad de Paths con `os.path`
Windows usa `\` mientras que Linux y macOS usan `/` para separar carpetas.
Para evitar incompatibilidades o errores por secuencias de escape (`\n`, `\t`), se debe usar siempre `os.path.join(...)` para construir rutas dinámicamente según el sistema operativo.

````python
# Para obtener la ruta segura, independiente del sistema operativo, usamos os.path.join(). 

# Los argumentos a entregar debe ser la ruta sin los \ o /, en forma de string.



ruta = os.path.join(".","archivos_2","carpeta_1",'carpeta_2')
# El . es para que nos dé el path relativo.

print(ruta)

# Nos entregará la ruta en formato de nuestro sistema operativo
````

### 4. Navegación por el Sistema de Archivos
- `os.listdir(ruta)`: Retorna una lista con los elementos presentes en **un solo nivel** dentro de ese directorio.
- `os.walk(ruta, topdown=True)`: Recorre un directorio de forma **recursiva**, retornando en cada iteración la `raiz`, los `directorios` y los `archivos`.

````python
ruta = os.path.join("archivos_2")

print(os.listdir(ruta))
print()
# Nos muestra todos los elementos de archivos_2

# Para recorrer el archivo y obtener su path, iteramos sobre os.walk(ruta, topdown = True)

for raiz, directorios, archivos in os.walk(ruta, topdown = True):
    # raiz (string) es la ruta en donde estamos parados
    # directorios (lista) son las subcarpetas ubicadas en raiz
    # archivos (lista) son todos los archivos presentes en raiz
    
    for archivo in archivos:
        actual = os.path.join(raiz,archivo)
        print(f"Estoy en {actual}")
````

### 5. `pathlib` (Enfoque Orientado a Objetos)
Es la alternativa moderna a `os.path`. Ofrece la clase `Path` con métodos integrados como:
- `.name`: Nombre completo del archivo (`basename`).
- `.suffix`: Extensión del archivo.
- `.is_file()`, `.is_dir()`, `.exists()`: Verificaciones del elemento.
- `.read_text()` / `.write_text()`: Lectura y escritura rápida.
- `.iterdir()`: Itera sobre los contenidos del directorio.

````python
from pathlib import Path
# Importamos la clase Path

ruta = Path("archivos_2")
# Creamos la ruta usando Path
print(ruta)
````

````python
# En el caso de tener una ruta más larga
ruta_1 = Path("archivos_2") / "carpeta_1" / "carpeta_2" / "ejemplo_1.txt"

print(ruta_1)
````

````python
# Volvemos a ruta = Path("archivos_2")
ruta = Path("archivos_2")

print(ruta.name) # Nombre del archivo (carpeta en este caso)
print(ruta.suffix) # Extensión (no hay extensión, ya que es una carpeta)
print(ruta_1.suffix)
print(ruta.is_file()) # Es archivo : False
print(ruta.is_dir()) # Es carpeta : True
for route in ruta.iterdir():
    print(f"Estoy en {route}")
# Ésta última funciona como os.listdir()
````

## `Terminal` y Ejecución de Python

La **terminal o consola** proporciona una interfaz de texto para darle instrucciones directamente al sistema operativo y interactuar con el sistema de archivos o ejecutar programas.

Sirve para:
* Poder abrir más de una terminal, y por ende, múltiples archivos simultáneamente.
* Saber exactamente nuestro path.

### 2. Tabla Resumen de Comandos Fundamentales

Todo comando tiene el siguiente formato `comando argumentos* -opciones* parámetros*`

| Acción / Propósito | Linux / macOS | Windows (cmd) |
| :--- | :--- | :--- |
| **Ver directorio actual** (*Print Working Directory*) | `pwd` | `cd` |
| **Cambiar de directorio** (*Change Directory*) | `cd <ruta>` | `cd <ruta>` |
| **Listar contenido del directorio** | `ls` | `dir` |
| **Crear un nuevo directorio** | `mkdir <nombre>` | `mkdir <nombre>` |
| **Crear un archivo vacío / actualizar marca de tiempo** | `touch <archivo>` | `echo "" > <archivo>` *(o `type nul > <archivo>`)* |
| **Ejecutar interprete / script de Python** | `python3 [script.py]` | `python [script.py]` |

---

### 3. Ejecución de Python y Manejo de Rutas

- **Modo interactivo:** Escribir `python3` (o `python` en Windows) sin argumentos abre la consola interactiva. Se sale ejecutando `exit()`.
- **Ejecución de scripts:** `python3 <nombre_archivo.py>` o especificando la ruta al archivo (`python3 ruta/al/archivo.py`).
- **Precaución con Rutas Relativas:** Al ejecutar un script desde una carpeta diferente a donde se encuentra ubicado, el directorio de trabajo activo (`os.getcwd()`) será donde está posicionada la consola en ese momento. Esto provocará un error de tipo `FileNotFoundError` si el script intenta abrir archivos locales mediante rutas relativas simples.


## `Git`

Git es un sistema distribuido de **control de versiones**. Permite registrar la evolución de un proyecto y conservar distintas versiones de sus archivos.


La idea general es llevar los cambios desde nuestra carpeta de trabajo hasta el repositorio remoto mediante una secuencia de comandos.

Un `commit` representa una versión del proyecto en un momento determinado. Funciona como una **foto (snapshot) del estado completo del proyecto** en ese momento. No significa que el commit contenga únicamente los archivos que se modificaron o agregaron en ese instante.

Por ejemplo:

```text
Commit 1 → main.py
Commit 2 → main.py + utils.py
Commit 3 → main.py + utils.py (con cambios)
Commit 4 → main.py + utils.py + README.md
```

## Comandos básicos

| Objetivo | Comando | Función |
|---|---|---|
| Obtener un repositorio por primera vez | `git clone` | Crea una copia local del repositorio remoto |
| Actualizar el repositorio local | `git pull` | Trae cambios desde el remoto |
| Preparar cambios | `git add` | Selecciona cambios para el próximo commit |
| Crear una versión | `git commit` | Registra un nuevo estado en el repositorio local |
| Subir versiones | `git push` | Publica commits locales en el repositorio remoto |

> Nota: `add` prepara cambios → `commit` crea una versión local → `push` la publica en el remoto; `pull` trae cambios del remoto y `clone` crea inicialmente la copia local.
